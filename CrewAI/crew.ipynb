```python
import os
from dotenv import load_dotenv
load_dotenv()

groq_api_key= os.getenv("GROQ_API_KEY")
```

```python
from crewai import Crew, Agent, Task, LLM

# By deafuls picks an open ai
llm = LLM(
    model="groq/llama-3.3-70b-versatile",
    api_key=groq_api_key,
)
```

```python
# create Agent
researcher = Agent(
    name="Researcher",
    description="An AI agent that researches and gather information on provided topic",
    goal="Find relavant information on given topic",
    role="Researcher",
    backstory="An AI Researcher loves gathering information",
)
```

```python
# create task
research_task = Task(
    name="Research Task",
    description="search for latest advancements in AI and summarize them",
    agent=researcher,
    expected_output="A Summary of latest advancements in AI"
)
```

```python
# Initialize Crew
crew = Crew(agents=[researcher], tasks=[research_task])
crew.kickoff()
```

```python
# Writer Agent
writer = Agent(
    name="Writer",
    description="Write a LinkedIn post based on provided research",
    goal="create structured linkedin post from gathered research data",
    role="Writer",
    backstory="An AI assistant designed for linkedin post that grab attention and provide some value",
)

# Reviewer Agent
reviewer = Agent(
    name="Reviewer",
    description="An AI editor that reviews linkedin post",
    goal="Ensure clarity, grammar and structure of a LinkedIn post",
    role="Reviewer",
    backstory="An AI reviewer to review linkedin post"
)
```

```python
gather_info = Task(
    name="Gather Information",
    description="Find the latest research paper and summarize key findings",
    agent=researcher,
    expected_output="In depth research of a provided topic",
)
```

```python
write_post = Task(
    name= "Post Writer",
    description="Writes LinkedIn post based on provided findings",
    agent=writer,
    expected_output="A structured and attention capturing linkedin post from a provided findings",
)
```

```python
review_post = Task(
    name="Review Linkedin Post",
    description="check linkedin post for structure, grammar and clarity",
    agent=reviewer,
    expected_output="A well reviewd linkedin post"
)
```

```python
multi_agent_crew = Crew(
    agents=[researcher, writer, reviewer],
    tasks=[gather_info, write_post, review_post]
)
```

```python
multi_agent_crew.kickoff()
```
