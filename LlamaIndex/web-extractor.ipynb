```python
import requests
from bs4 import BeautifulSoup, Tag

url = "https://supercell.com/en/parents/#I'VE-MADE-AN-IN-GAME-PURCHASE,-BUT-I-HAVEN'T-RECEIVED-WHAT-I-PAID-FOR.-WHAT-NOW?"


response = requests.get(url)
soup = BeautifulSoup(response.content, "html.parser")

body = soup.body
```

```python
current_question = None
current_answers = []

questions = []
answers = []
```

```python
for tag in body.descendants:
    if isinstance(tag, Tag):
        # Iterate Over all Questions
        if tag.name == "h3":
            if current_question and current_answers:
                questions.append(current_question)
                answers.append(" ".join(current_answers))

            # Start a new question-answer pair
            current_question = tag.get_text(strip=True)
            current_answers = []
        # Add Answers which are present inconsecutive html tags
        elif tag.name == "p" and current_question:
            current_answers.append(tag.get_text(strip=True))


if current_question and current_answers:
    questions.append(current_question)
    answers.append(" ".join(current_answers))
```

```python
print(len(questions))
print(len(answers))
```

```python
import pandas as pd

df = pd.DataFrame(
    {
        "Questions": questions,
        "Answers": answers
    }
)
```
